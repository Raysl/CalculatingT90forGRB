```python
import batanalysis as ba
import swiftbat
from swifttools.swift_too import ObsQuery 

import matplotlib.pyplot as plt
import numpy as np
import scipy as sp
from pathlib import Path
from astropy.io import fits
from astropy.time import Time, TimeDelta
import astropy.units as u
from astropy.coordinates import SkyCoord
import datetime
import os
from swifttools.swift_too import GUANO
from swifttools.swift_too import Data
from astropy.stats import bayesian_blocks  
```

# Example

```python
oq = ObsQuery(begin="2023-04-09 04:56:56", length=0.1)
oq
```

```python
data = Data(obsid=oq[1].obsid, bat=True, outdir="~/Downloads/", clobber=True) #Collecting the data
```

```python
event=ba.BatEvent(oq[1].obsid,obs_dir = "/storage/home/zbl5364/Downloads/", is_guano=True, recalc = True, ra = 329.53*u.deg, dec = 52.85*u.deg)
# Downloading the data
```

```python
lc = event.create_lightcurve(timedelta=np.timedelta64(16, "100ms"), recalc = True)#Set up the light curve
```

```python
lc.set_timebins(timebinalg = "bayesian", save_durations = True)# Changing the bins
```

```python
total_duration = (lc.tdurs['T90']['TSTOP'] - lc.tdurs['T90']['TSTART']).to('s').value
print("Total duration:", total_duration)
```
